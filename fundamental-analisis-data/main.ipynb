## Exploratory Data Analysis

### Bagaimana trend perentalan bulanan jika dilihat dari perental member dan non-member?

```python
# code here
day_df.head()
```

```python
# Mencari trend untuk tahun
print("Treng Perentalan (Tahun): ")
all_year = day_df.groupby("yr").agg({
    'casual':'sum',
    'registered':'sum',
    'cnt':'sum'
}).reset_index()

all_year.sort_values(by='cnt', ascending=False)
```

```python
year_1 = day_df[day_df['yr']==0]
year_2 = day_df[day_df['yr']==1]
```

```python
print("Trend perentalan bulanan tahun pertama: ")
year1 = year_1.groupby("mnth").agg({
    'casual':'sum',
    'registered':'sum',
    'cnt':'sum'
}).reset_index()

year1.sort_values(by='cnt', ascending=False)
```

```python
print("Trend perentalan bulanan tahun kedua: ")
year2 = year_2.groupby("mnth").agg({
    'casual':'sum',
    'registered':'sum',
    'cnt':'sum'
}).reset_index()

year2.sort_values(by='cnt', ascending=False)
```

```python
print("Trend perentalan bulanan semua tahun: ")
allyear = day_df.groupby("mnth").agg({
    'casual':'sum',
    'registered':'sum',
    'cnt':'sum'
}).reset_index()

allyear.sort_values(by='cnt', ascending=False)
```

### Bagaimana trend perentalan antara weekend dan weekdays?

```python
# code here
print("Trend perentalan antara weekend dan weekdays: ")
week_rent = day_df.groupby("workingday").agg({
    'casual':'sum',
    'registered':'sum',
    'cnt':'sum'
}).reset_index()

week_rent.sort_values(by='cnt', ascending=False)
```

### Di jam-jam berapakah perentalan ramai dan sepi?

```python
# code here
hour_df.head()
```

```python
# code here
print("Trend perentalan berdasarkan jam: ")
hourly_rent = hour_df.groupby("hr").agg({
    'casual':'sum',
    'registered':'sum',
    'cnt':'sum'
}).reset_index()

hourly_rent.sort_values(by='cnt', ascending=False)
```
